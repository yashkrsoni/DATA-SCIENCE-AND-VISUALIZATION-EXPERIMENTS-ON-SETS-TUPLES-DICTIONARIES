#Q7.5 Write swap_message(val1, val2) to swap and print local variables, then show that the original variables remain unchanged.

```python
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)
x = 6
y = 3
swap_message(x, y)
print("Outside function:")
print("x =", x)
print("y =", y)
```

```python
counter = 0
def increment_counter():
    global counter
    counter = counter + 1
increment_counter()
increment_counter()
print(counter)
```
