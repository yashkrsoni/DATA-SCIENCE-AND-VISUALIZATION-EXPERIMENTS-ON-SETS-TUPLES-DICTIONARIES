```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer, KNNImputer
```

#Q1. Create DataFrame of 10 students

```python
df = pd.DataFrame({
    'Student': list('ABCDEFGHIJ'),
    'Age': [18, 19, np.nan, 21, 20, np.nan, 22, 19, 21, 20],
    'Marks': [85, np.nan, 78, 92, 88, 75, np.nan, 80, 90, np.nan],
    'City': ['Delhi', 'Agra', np.nan, 'Mathura', 'Agra',
             'Delhi', np.nan, 'Mathura', 'Delhi', np.nan]
})
print("1. Original Dataset:")
print(df.to_string(index=False))
```

#Q2.  Count missing values and percentage

```python
print("\n2. Missing Values Count:")
print(df.isnull().sum())
print("\nPercentage of Missing Values:")
print((df.isnull().mean() * 100).round(2))
```

#Q3. Drop missing rows

```python
drop_df = df.dropna()
print("\n3. Dataset After dropna():")
print(drop_df.to_string(index=False))
print("Records remaining:", len(drop_df))
# Mean imputation
mean_df = df.copy()
mean_df['Age'] = mean_df['Age'].fillna(mean_df['Age'].mean())
mean_df['Marks'] = mean_df['Marks'].fillna(mean_df['Marks'].mean())
# Median imputation
median_df = df.copy()
median_df['Age'] = median_df['Age'].fillna(median_df['Age'].median())
median_df['Marks'] = median_df['Marks'].fillna(median_df['Marks'].median())
print("\nMean Imputation:")
print(mean_df[['Student', 'Age', 'Marks']].round(2).to_string(index=False))
print("\nMedian Imputation:")
print(median_df[['Student', 'Age', 'Marks']].to_string(index=False))
```

#Q4. Most frequent imputation for City

```python
city_imputer = SimpleImputer(strategy='most_frequent')
df_city = df.copy()
df_city[['City']] = city_imputer.fit_transform(df_city[['City']])
print("\n4. City After Most Frequent Imputation:")
print(df_city[['Student', 'City']].to_string(index=False))
print("Most frequent city:", city_imputer.statistics_[0])
```

#Q5.  KNN Imputation with k=2 and k=3

```python
numeric_cols = ['Age', 'Marks']
knn2 = KNNImputer(n_neighbors=2)
knn3 = KNNImputer(n_neighbors=3)
result2 = pd.DataFrame(
    knn2.fit_transform(df[numeric_cols]),
    columns=numeric_cols
)
result3 = pd.DataFrame(
    knn3.fit_transform(df[numeric_cols]),
    columns=numeric_cols
)
print("\n5. KNN Imputation (n_neighbors=2):")
print(result2.round(2).to_string(index=False))
print("\nKNN Imputation (n_neighbors=3):")
print(result3.round(2).to_string(index=False))
print("\nEstimated values at missing positions:")
for col in numeric_cols:
    missing_rows = df[col].isnull()
    print(f"\n{col}:")
    print(pd.DataFrame({
        'Student': df.loc[missing_rows, 'Student'].values,
        'KNN_k2': result2.loc[missing_rows, col].round(2).values,
        'KNN_k3': result3.loc[missing_rows, col].round(2).values
    }).to_string(index=False))
```

#Q6. Extract rows containing missing values

```python
missing_rows = df[df.isnull().any(axis=1)]
print("\n6. Rows Containing Missing Values:")
print(missing_rows.to_string(index=False))
print("Rows containing missing values:", len(missing_rows))
```
