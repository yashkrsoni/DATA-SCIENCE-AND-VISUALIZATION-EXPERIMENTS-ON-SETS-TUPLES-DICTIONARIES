```python
cities = {"Delhi", "Agra"}
print(cities)
print(type(cities))
```

```python
cities.add("Mathura")
print(cities)
cities.update(["Lucknow", "Kanpur"])
print(cities)
```

```python
cities.remove("Agra")
```

```python
print(cities)
cities.discard("Lucknow")
print(cities)
```

```python
cities.discard("Agra")
```

```python
cities.discard("Delhi")
print(cities)
```

```python
cities = {"London", "Paris", "New York"}
print(cities)
rem = cities.pop()
print(rem)
print(cities)
```

```python
cities.clear()
print(cities)
```

```python
A = {"Delhi", "Agra", "Mathura"}
B = {"Agra", "Delhi", "Lucknow"}
print(A|B) # both are same
```

```python
print(A.union(B)) # both are same
```

```python
print(A & B)
```

```python
print(A - B)
```

```python
print(A ^ B) # sign of symmetric difference
```
