#1. Age Pass/Fail + Fare High/Low using Pandas

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Riya", "Neha", "Karan"],
    "Age": [20, 16, 25, 17, 30],
    "Fare": [600, 300, 450, 700, 250]
})

print(df)

# Pass/Fail
df["Result"] = df["Age"].apply(lambda x: "Pass" if x >= 18 else "Fail")

# Fare High/Low
df["Fare_Type"] = df["Fare"].apply(lambda x: "High" if x >= 500 else "Low")

print("\nFinal DataFrame:")
print(df)
```

#2. Create 5 names + Insert, Update, Remove + Length, Minimum, Maximum, Total

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Riya", "Neha", "Karan"]
})

print("Original DataFrame:")
print(df)

# Insert
df.loc[len(df)] = ["Yash"]
print("\nAfter Insert:")
print(df)

# Update
df.loc[df["Name"] == "Aman", "Name"] = "Arjun"
print("\nAfter Update:")
print(df)

# Remove
df = df[df["Name"] != "Neha"]
print("\nAfter Remove:")
print(df)

# Length
print("\nLength:", len(df))

# Minimum
print("Minimum:", df["Name"].min())

# Maximum
print("Maximum:", df["Name"].max())

# Total
print("Total:", df["Name"].count())
```

#3. Tuple of ages + access, update, combine

```python
import pandas as pd

ages = (18, 20, 22, 25, 30)

df1 = pd.DataFrame({"Age": ages})

print("Original DataFrame:")
print(df1)

# Access elements
print("\nFirst Age:", df1.iloc[0]["Age"])
print("Third Age:", df1.iloc[2]["Age"])

# Update
df1.loc[1, "Age"] = 21

print("\nAfter Update:")
print(df1)

# Another tuple
ages2 = (35, 40, 45)

df2 = pd.DataFrame({"Age": ages2})

# Combine
combined = pd.concat([df1, df2], ignore_index=True)

print("\nCombined DataFrame:")
print(combined)
```

#4. Set: Remove duplicates + Union + Intersection of passenger IDs

```python
import pandas as pd

set1 = {101, 102, 103, 104, 104, 102}
set2 = {103, 104, 105, 106, 106}

# Remove duplicates
df1 = pd.DataFrame({"Passenger_ID": list(set1)})
df2 = pd.DataFrame({"Passenger_ID": list(set2)})

print("Set 1:")
print(df1)

print("\nSet 2:")
print(df2)

# Union
union = pd.concat([df1, df2]).drop_duplicates().reset_index(drop=True)

print("\nUnion:")
print(union)

# Intersection
intersection = pd.merge(df1, df2, on="Passenger_ID")

print("\nIntersection:")
print(intersection)
```

#5. Dictionary: Name, Age, Fare, Family Size + Update Age + Delete Fare

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Aman", "Riya"],
    "Age": [25, 20, 22],
    "Fare": [750, 500, 600],
    "Family_Size": [4, 3, 5]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Update Age
df.loc[df["Name"] == "Rahul", "Age"] = 26

print("\nAfter Updating Age:")
print(df)

# Delete Fare
df = df.drop("Fare", axis=1)

print("\nAfter Deleting Fare:")
print(df)
```

#6. List of passenger dictionaries + Adult passengers + DataFrame

```python
import pandas as pd

passengers = [
    {"Name": "Rahul", "Age": 25},
    {"Name": "Aman", "Age": 16},
    {"Name": "Riya", "Age": 22},
    {"Name": "Neha", "Age": 15},
    {"Name": "Karan", "Age": 30}
]

# Convert list of dictionaries into DataFrame
df = pd.DataFrame(passengers)

print("Passenger DataFrame:")
print(df)

# Adult passengers
adult_passengers = df[df["Age"] >= 18]

print("\nAdult Passengers:")
print(adult_passengers)

# Name-Age pair list
pair_list = list(zip(df["Name"], df["Age"]))

print("\nName-Age Pair List:")
print(pair_list)

# Convert pair list into DataFrame
pair_df = pd.DataFrame(pair_list, columns=["Name", "Age"])

print("\nFinal DataFrame:")
print(pair_df)
```
