```python
import numpy as np
import pandas as pd
```

#Sample Dataset

```python
df = pd.DataFrame({
    'Student_ID': [101, 102, 103, 103, 104],
    'Name': ['Aman', 'Riya', 'Rahul', 'Rahul', 'Priya'],
    'Marks': [80, np.nan, 70, 70, 90],
    'Status': ['Pass', 'NA', '', 'Pass', 'Pass']
})

print("Original Dataset:")
print(df)
```

## Q1. Convert blank strings and 'NA' into missing values

```python
df['Status'] = df['Status'].replace(['', 'NA'], np.nan)
print("\nQ1. After replacing blank strings and NA:")
print(df)
```

## Q2. Compare dropping missing Marks with filling median

```python
drop_df = df.dropna(subset=['Marks'])
median_df = df.copy()
median = median_df['Marks'].median()
median_df['Marks'] = median_df['Marks'].fillna(median)
print("\nQ2. After dropping rows with missing Marks:")
print(drop_df)
print("\nAfter filling missing Marks with median:")
print(median_df)
```

## Q3. Why deleting repeated Student_ID values can be wrong

```python
print("\nQ3. Duplicate Student_ID records:")
print(df[df.duplicated(subset=['Student_ID'], keep=False)])
# Remove only exact duplicate rows if needed
unique_df = df.drop_duplicates()
print("\nAfter removing exact duplicate rows:")
print(unique_df)
```

## Q4. Difference between NaN and NaT

```python
data = pd.DataFrame({
    'Marks': [80, np.nan, 90],
    'Date': pd.to_datetime(
        ['2026-01-10', None, '2026-01-12']
    )
})
print("\nQ4. NaN and NaT example:")
print(data)
print("\nNaN means missing numeric/general data.")
print("NaT means missing date or time data.")
```
