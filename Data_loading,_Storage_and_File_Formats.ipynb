```python
import pandas as pd

data = {
   "Name": ["Tej","Yash","Yogesh","Yuvraj","Shubham"],
  "Age" : [20,21,20,22,21],
   "Department" : ["CSE", "IT","CSE","ECE","CSE"],
   "Marks" : [85,90,87,91,88]}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.tail())
print("Shape:",df.shape)
print("Columns:",df.columns.tolist())
print("\nData types")
print(df.dtypes)
print("\nInformation:")
df.info()
print("\nStatistics:")
print(df.describe())
```

```python
df.to_csv("student_output.csv", index = False)
```

```python
iris_df = pd.read_csv("iris.csv")
print(iris_df.head())
```

```python
print("Shape:",iris_df.shape)
print("Columns:",iris_df.columns.tolist())

print("\nData types")
print(iris_df.dtypes)

print("\nMissing values:")
print(iris_df.isnull().sum())

print("\nInformation:")
iris_df.info()

print("\nStatistics:")
print(iris_df.describe(include='all'))
```
